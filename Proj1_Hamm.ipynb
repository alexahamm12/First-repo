4.  Show Python code  in the following code box which prints the following statistics of the bill length of all of the penguins, broken down by the species:  minimum, median, mean and maximum


```python
ss = df.groupby('species').agg({'bill_length_mm':['min','median','mean','max']})
```

5.  Repeat the above, but sort the list by the minimum bill length in ascending order.  Hint:  You may need to figure out exactly what the columns names are to sort by, and will have to do some research on how to do this.



```python
ss = df.groupby('species').agg({'bill_length_mm':['min','median','mean','max']})
ss.sort_values(('bill_length_mm', 'min'), ascending=True)

                                                                                         
```
